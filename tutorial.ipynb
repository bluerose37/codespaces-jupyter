```python
#multiple conditions
#LOGICAL OPERATORS
# AND,OR operators--not only for python but,
#for all lang that support if-else
#AND-both should be true,OR-atleast one,NOT

has_high_income=False
has_good_credits=True

if has_high_income or has_good_credits:
    print("Eligible for loan")
```

```python
# NOT
has_good_credits=True
has_criminal_record=True

if has_good_credits and not has_criminal_record:
    print("Eligible for loan")
```
