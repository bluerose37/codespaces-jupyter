```python
#Strings
course='Python for Beginners'
print(len(course)) #len function--general purpode func.

#when function is belonged to a object---is called method

print(course.upper()) #is method
print(course) #upper() doesn't modify original string
print(course.lower())
print(course.find('o')) #print the index of the string
print(course.find('O')) #-1 cuz cap'O' doesn't exist
print(course.find('Beginners'))
#find() is case sensitive to upper and lower case

print(course.replace('beginners','Absolute Beginners'))
#replace()--not case sensitive--works for both'B' and 'b'

# 'in' operator--case sensitive
print('Pyt' in course)
print('python' in course)

'''
len()
course.upper()
course.lower()
course.title()
course.find()
course.replace()
'...' in course --to find whether characters are in string
'''
```
